

## Post-Doppler Adaptive Processing: Derivation



The goal of adaptive processing is to combine the outputs of $K$ receive subarrays so that a target at a known angle and Doppler is preserved, while clutter and noise arriving from other angles are suppressed as much as possible. We outline below how this combination is derived, starting from the raw per-subarray, per-pulse signal model, compressing to a single Doppler bin, and arriving at the constrained minimization that defines the adaptive weights.



### Signal model



Consider subarray $k$, pulse $m$ ($m\in\{0,\dots,M-1\}$), at a single range gate $\tau$. The received signal is a sum of contributions from $N$ clutter patches, one target, and additive noise:


$x_{k,m}(t) = \sum_{i=1}^{N} \sqrt{RCS_i}\cdot AF(\theta_i-\theta_{tx})\cdot s(t-\tau)\cdot e^{j\phi_k(\theta_i-\theta_{rx})}\, e^{j2\pi f_{d,i}\,m\,T}$

$
{}+ \sqrt{RCS_t}\cdot AF(\theta_t-\theta_{tx})\cdot s(t-\tau)\cdot e^{j\phi_k(\theta_t-\theta_{rx})}\, e^{j2\pi f_{d,t}\cdot m\cdot T} + n_{k,m}(t)
$

<br>
Each clutter patch $i$ and the target contribute a transmit pattern factor $AF(\theta_i-\theta_{tx})$, the compressed waveform $s(t-\tau)$, a subarray-dependent receive phase $e^{j\phi_k(\theta_i-\theta_{rx})}$, and a pulse-to-pulse Doppler phase $e^{j2\pi f_{d}\,m\,T}$. All clutter patches and the target are assumed to lie at the same range $\tau$; $n_{k,m}(t)$ is additive receiver noise, uncorrelated across $k$ and $m$.



### Doppler compression to a single bin

<br>
To limit processing to a Doppler bin $f_D = \dfrac{k_d}{MT}$, $k_d\in\{0,\dots,M-1\}$, the $M$ pulses are coherently combined:

$z_k(t) = \sum_{m=0}^{M-1} x_{k,m}(t)\, e^{-j2\pi f_D\, m\, T}$


<br>
Substituting the signal model and exchanging the order of summation gives:

$
z_k(t) = \sum_{i=1}^{N}\sqrt{RCS_i}\cdot AF(\theta_i-\theta_{tx})\cdot s(t-\tau)\cdot e^{j\phi_k(\theta_i-\theta_{rx})}\sum_{m=0}^{M-1} e^{j2\pi (f_{d,i}-f_D)\,m\,T}
$

${}+\sqrt{RCS_t}\cdot AF(\theta_t-\theta_{tx})\cdot s(t-\tau)\cdot e^{j\phi_k(\theta_t-\theta_{rx})}\sum_{m=0}^{M-1} e^{j2\pi (f_{d,t}-f_D)\,m\,T} + n_k(t)$

<br>
where $n_k(t) = \sum_{m=0}^{M-1} n_{k,m}(t)\,e^{-j2\pi f_D m T}$ is the compressed noise term.
<br>Given an initial estimate of the target Doppler, setting $f_D=f_{d,t}$ collapses the target's inner sum to $M$ (constructive, on-bin), while each clutter term retains its Dirichlet-kernel weighting $\sum_m e^{j2\pi(f_{d,i}-f_{d,t})mT}$ — the residual leakage of off-bin clutter into the target's bin.


### Objective function
<br>
With the target isolated in its own Doppler bin, the remaining problem is purely spatial: find combining weights $w_k$ across the $K$ subarrays that minimize the combined clutter-plus-noise power at the output, while preserving unit gain toward the target:

$
\min_{\{w_k\}} \; E\Big[\Big|\sum_{k=1}^{K} w_k^{*}\, z_{k,\text{clutter+noise}}(t)\Big|^2\Big]
$
<br> subject to
$
\sum_{k=1}^{K} w_k^{*}\, p_{T,k} = 1
$
<br>
i.e. $\mathbf{w}^H\mathbf{p}_T=1$, where $p_{T,k}=e^{j\phi_k(\theta_t-\theta_{rx})}$ is the target's spatial steering value and $w_k^{*}$ denotes the conjugate of the combining weight (this convention is used consistently in both the constraint and the objective).
<br>
This is the standard MVDR/Capon problem, solved by $\mathbf{w} = \dfrac{R^{-1}\mathbf{p}_T}{\mathbf{p}_T^H R^{-1}\mathbf{p}_T}$, where $R$ is the covariance of $z_{k,\text{clutter+noise}}(t)$ across subarrays.

$
\mathcal{L}(\mathbf{w},\lambda) = \mathbf{w}^H R\,\mathbf{w} + \lambda\big(1-\mathbf{w}^H\mathbf{p}_T\big) $

<br><br>
Setting the gradient to zero gives $R\mathbf{w}=\lambda \mathbf{p}_T \Rightarrow \mathbf{w}=\lambda R^{-1}\mathbf{p}_T$, and enforcing the constraint fixes $\lambda$, reproducing $\mathbf{w}_{\text{adapt}}=\dfrac{R^{-1}\mathbf{p}_T}{\mathbf{p}_T^H R^{-1}\mathbf{p}_T}$ — the same closed form, but arrived at as the unique minimizer of output power, not as a maximizer of a ratio.

## SINR Improvement from Adaptive Processing
<br>
The benefit of solving for $\mathbf{w}$ rather than using a fixed (conventional) combiner is measured by the output signal-to-interference-plus-noise ratio (SINR).

### Output SINR for a general weight vector
<br>
For any combining weights $\mathbf{w}$ applied to $\mathbf{z}(t) = [z_1(t),\dots,z_K(t)]^T$, the output is $y(t)=\mathbf{w}^H\mathbf{z}(t)$, and the output SINR is:

$
\text{SINR}(\mathbf{w}) = \frac{\big|\mathbf{w}^H\mathbf{p}_T\big|^2 \, RCS_t}{\mathbf{w}^H R\, \mathbf{w}}
$
<br>
Since, by design, $\mathbf{w}^H\mathbf{p}_T=1$ and ${\mathbf{w}^H R \mathbf{w}}$ is minimized, adaptive processing guarantees maximum $\text{SINR}$.
